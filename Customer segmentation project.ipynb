# KMeans Clustering 

```python
# use elbow method to find the best cluster
```

```python
x = dataset[["Age","Annual Income (k$)","Spending Score (1-100)","en_gender"]]
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []

for i in range(2,21):
    km = KMeans(n_clusters=i, init="k-means++")
    km.fit(x)
    wcss.append(km.inertia_)
```

```python
plt.plot([i for i in range(2,21)], wcss, marker="o")
plt.xlabel("no_of_customer")
plt.xticks([i for i in range(2,21)])
plt.ylabel("WCSS")
plt.grid(axis="x")
plt.show()
```

```python
kn = KMeans(n_clusters=2, init="k-means++")
dataset["predict"] = kn.fit_predict(x)
```

```python
sns.pairplot(data=dataset, hue="predict")
plt.show()
```
